# CanchaMatch

_Note_: The context and dataset was created by AI, for making more challenging. The AI was the CEO of the company throughout the development

## Company

CanchaMatch is a tennis clubs' administrator in Colombia. Users reserve the court through its mobile app. Usually, the app receives around 1.000 reserves per day. The company has Clay courts, hard courts and synthetic courts. Furthermore, it offers a tennis coach, racket rental and ball machine rental.

## Context

During the last few months, the company has noticed a weird reduction in income. After checking the financial reports, they found out it may happen because reservations aren’t giving the ideal return. Some employees thought that reservations incremented during the last few months, so it's strange looking the decreasing in the income. They don’t know:

- Why it happen
- If all locations have the same problem
- What kind of clients have this behaviour

The boss said: “We want to understand what has happened. If it’s necessary, create a model that helps us to anticipate these changes”

## Extra information

This information was gotten by asking questions to the CEO:

###  Payments
The client must pay before the reservation's day

### Memebrship system
There are 2 memberships:
1. Gold: The client has 25% of discount in reservations, has the most important priority to make reservations, discounts with machine ball rental and 2 canelations without penalization
2. Silver: The client has 10% of discount in reservations, has priority to make reservations, and 1 canelation without penalization

If a client doesn't have a membership, the prices will be higher for them and doesn't have discounts

> Note: the dataset doesn't have the incomes that come from the memberships

### The level of the player
At first it's defined by the client. It's possible that this data change if a coach identify a different level

### Cancellations
When a client cancel  reservation, the company refunds a percentage according with the moment of cancelations

| **Cancellation moment** | **Percentage of Refund** |
|----|-----|
| 24 hours before the reservation | 100% |
| Between 6 to 24 hours before reservation | 50% |
| Lees than 6 hours before reservations | 0% |

The refund given is in the variable `refund_amount` and this only has the cancellation refund.

The income depends on `price_paid - refund_amount`


## Question to answer

### Essential question
> Why is the company reducing incomes? 

### Other questions
> Does it happen the same at all locations?

> Do all clients have the same behaviour?

> Are reservations decreasing?

> What do the customer prefer: pay only a court or pay court and other services?

> Are there many customers with memberships? Do the discount received affect our incomes?

## Hypothesis
- If the cancellations are increased, then the income has decreased
- If the reservations hours are less, the income has decreased
- If the customer doesn't rent the coach and ball machine, the income has decreased
